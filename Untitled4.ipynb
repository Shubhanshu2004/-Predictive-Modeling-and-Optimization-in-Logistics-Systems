### Problem Definition and Data Simulation

To begin, we will simulate a dataset representing typical logistics delivery operations. This hypothetical dataset will contain features that are commonly associated with delivery times, allowing us to build a predictive model. The target variable will be `delivery_time_minutes`.

```python
import pandas as pd
import numpy as np

# Set a random seed for reproducibility
np.random.seed(42)

# Number of delivery records to simulate
n_samples = 1000

# 1. Simulate Features
order_ids = [f'ORD-{i+1:04d}' for i in range(n_samples)]

# Distance in kilometers (e.g., 5 km to 100 km)
distance_km = np.random.uniform(5, 100, n_samples)

# Traffic level (categorical)
traffic_levels = np.random.choice(['low', 'medium', 'high'], n_samples, p=[0.5, 0.3, 0.2])

# Weather condition (categorical)
weather_conditions = np.random.choice(['clear', 'rainy', 'snowy'], n_samples, p=[0.7, 0.2, 0.1])

# Day of the week (0=Monday, 6=Sunday)
day_of_week = np.random.randint(0, 7, n_samples)

# Time of day (hour of delivery start, e.g., 8 AM to 8 PM)
time_of_day_hour = np.random.randint(8, 20, n_samples)

# 2. Simulate Target Variable: Delivery Time in Minutes
# Base delivery time (linear relationship with distance)
delivery_time_minutes = 2 * distance_km # 2 minutes per km as a base

# Add effects of traffic
delivery_time_minutes += np.where(traffic_levels == 'medium', 10, 0)
delivery_time_minutes += np.where(traffic_levels == 'high', 25, 0)

# Add effects of weather
delivery_time_minutes += np.where(weather_conditions == 'rainy', 15, 0)
delivery_time_minutes += np.where(weather_conditions == 'snowy', 40, 0)

# Add effects of peak hours (e.g., 4 PM - 6 PM)
delivery_time_minutes += np.where((time_of_day_hour >= 16) & (time_of_day_hour <= 18), 10, 0)

# Add some random noise
delivery_time_minutes += np.random.normal(0, 10, n_samples) # Add Gaussian noise

# Ensure delivery time is not negative
delivery_time_minutes = np.maximum(delivery_time_minutes, 10) # Minimum 10 minutes

# Create DataFrame
data = pd.DataFrame({
    'order_id': order_ids,
    'distance_km': distance_km,
    'traffic_level': traffic_levels,
    'weather_condition': weather_conditions,
    'day_of_week': day_of_week,
    'time_of_day_hour': time_of_day_hour,
    'delivery_time_minutes': delivery_time_minutes
})

# Display the first 5 rows of the simulated data
display(data.head())

# Display data types and basic statistics
print(data.info())
display(data.describe())
```

### Data Exploration: Relationship between Distance and Delivery Time

We will now visualize the relationship between `distance_km` and `delivery_time_minutes` using a scatter plot. This helps us understand the underlying pattern and the strength of their correlation, which is crucial for predictive modeling.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a scatter plot to visualize the relationship between distance and delivery time
plt.figure(figsize=(10, 6))
sns.scatterplot(x='distance_km', y='delivery_time_minutes', data=data, alpha=0.6)
plt.title('Distance vs. Delivery Time')
plt.xlabel('Distance (km)')
plt.ylabel('Delivery Time (minutes)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### Data Preprocessing: One-Hot Encoding

To prepare the categorical features (`traffic_level` and `weather_condition`) for our predictive model, we will apply one-hot encoding. This process converts categorical variables into a numerical format, where each category becomes a new binary (0 or 1) column. This prevents the model from assuming any ordinal relationship between categories.

```python
# Perform one-hot encoding on 'traffic_level' and 'weather_condition'
data_encoded = pd.get_dummies(data, columns=['traffic_level', 'weather_condition'], drop_first=True)

# Display the first few rows of the encoded data
display(data_encoded.head())

# Display the new column information to show the effect of encoding
print(data_encoded.info())
```
